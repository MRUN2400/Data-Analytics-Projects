```python
df=df.drop('Discount Applied',axis=1)
```

```python
df.head()
```

```python
df.columns=df.columns.str.replace(' ','_')
```

```python
df.columns
```

```python
df.columns=df.columns.str.replace('Purchase_Amount_(USD)','Purchase_Amount_USD')
```

```python
df.columns
```

```python
df.isnull().sum()
```

```python
df['Review_Rating'] = df.groupby('Category')['Review_Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
label=['Young Adult','Adult','Middle','Old']
df['age_group'] = pd.qcut(df['Age'], q=4, labels=label)
```

```python
frequency_mapping={
    'Fortnightly':14,
    'Bi-Weekly':14,
    'Weekly':7,
    'Monthly':30,
    'Every 3 months':30,
    'Quarterly':90,
    'Annually':350
     
}
df['Purchase_Frequency']=df['Frequency_of_Purchases'].map(frequency_mapping)
```
